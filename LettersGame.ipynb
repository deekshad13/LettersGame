# Letters Game
*A fun word guessing game!*

```python
import random
import string

# Variables to adjust the gameplay 
words_file_path = 'safedict_full.txt' # file containing one word per line. Thanks and credit to https://github.com/InnovativeInventor/dict4schools
min_word_len = 15
letter_points_dict = {'A':1, 'E':1, 'I':1, 'O':1, 'U':1, 'L':1, 'N':1, 'S':1,
                     'T':1, 'R':1,'D':2, 'G':2,'B':3, 'C':3, 'M':3, 'P':3,
                     'F':4, 'H':4, 'V':4, 'W':4, 'Y':4, 'K':5, 'J':8, 'X':8,
                     'Q':10, 'Z':10}

def load_words():
    word_list = []
    # TODO: (2) Write code here to read in the words file.
    with open(words_file_path, 'r')as file:
        for line in file:
            word = line.strip('\n')
            if len(word) > min_word_len and '-' not in word:
                word_list.append(word)
    return word_list

def print_welcome():
    # TODO: (1) Print a welcome message here.
    print("Welcome to the word game!!")


def print_instructions():
    # TODO: (8) Print an instructions message here.
    print('HOW TO PLAY')
    print('1. The word is hidden. Each _ is a missing letter.')
    print('2. Type one letter per turn.')
    print('3. Correct letter = points added. Wrong letter = points taken away.')
    print('4. Rare letters (Q, Z, X, J) are worth more, both ways.')
    print('5. Reveal the whole word to finish. Type "quit" to leave early.')
    

def display_word(word, guess_list):
    # TODO: (5) write a loop which displays each character, display un-geuessed characters as underscores.
    for letter in word:
        if letter in guess_list:
            print(letter, end = ' ')
        else:
            print('_', end = ' ')
    print()
    

def is_letter(x):
    # make sure there is only 1 character and it in the set of a-z A-Z
    return len(x) == 1 and x in string.ascii_letters

def is_game_over(word, guess_list):
    # loop over each letter in the mystery word
    for letter in word:
        # is this an un-guessed letter?
        if letter not in guess_list:
            # the player has not guess this letter yet
            return False # game is not over
    return True # the above loop ran without hitting the False, hence all letters are guessed, game over

def main():    
    print_welcome()
    print_instructions()

    # load words and choose word one at random
    # TODO: (3) call you load words function and store the resultant list in variable
    words = load_words()
    # TODO: (4) randomly choose a word from you word list variable, you must call this "word"
    word = random.choice(words)

    # initialise game variables
    guess_list = []
    score = 0

    # main loop
    while(True):
        # display the game state
        # TODO: (6) call your display word function so the user can see there progress
        display_word(word, guess_list)
        print('You have guessed:' + str(guess_list))
        print('Your score is ' + str(score))

        # get user input
        # TODO: (7) get users input, ask the user to type a letter, call this 'user_input' (as seen below)
        user_input = input('Type a letter (or "quit" to exit):').lower()

        # based on input choose action to take
        if user_input == 'quit': # they typed quit
            # breaks out of the loop ending the game
            break
        elif is_letter(user_input):  # they typed a single letter
            # Have they already guessed this letter?
            if user_input in guess_list:
                print('You have already guessed this letter.')
            else:
                # does the letter they typed appear in the word
                if user_input in word:
                    # yes, increment score, according to the letter value
                    score = score + letter_points_dict[user_input.upper()]
                else:
                    # no, decrement score, according to the letter value
                    score = score - letter_points_dict[user_input.upper()]
                # record the letter
                guess_list.append(user_input)

            # check if they have guessed all letters
            if is_game_over(word, guess_list):
                print('The word was ' + word)
                print('Game Over, you score is: ' + str(score))

                # TODO: (9) based on their score display a message!
                if score < 0:
                    print('Better luck next time!')
                elif score < 10:
                    print('Not bad, you got there in the end!')
                elif score < 30:
                    print('Great job, well played!')
                else:
                    print('Amazing! You are a word wizard!')
                
                
                # breaks out of the loop ending the game
                break
            

main()
```
